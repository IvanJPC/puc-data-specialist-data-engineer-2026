# Catalog and Schemas creation
The Medallion Architecture is a data lakehouse design pattern that organizes data into layers based on quality and refinement level. The project will use this pattern

- Raw Layer (00-raw)

    Purpose: Initial landing zone for data exactly as ingested from source systems

    Characteristics:

        Unprocessed, unvalidated data
        Original structure and format preserved
        Immutable record of source data
        Often includes metadata like ingestion timestamps
        A list of volumes will receive the source data transfered and preserved in order to be refined through each layer  


        

```python
    %sql
    -- Defining the catalog and schema names 
    -- Schema names using number prefix to order by name
    DECLARE OR REPLACE catalog_name STRING DEFAULT 'puc_data_specialist_de_2026_09';
    DECLARE OR REPLACE schema_name_raw STRING DEFAULT '00-raw';
    DECLARE OR REPLACE schema_name_bronze STRING DEFAULT '01-bronze';
    DECLARE OR REPLACE schema_name_silver STRING DEFAULT '02-silver';
    DECLARE OR REPLACE schema_name_gold STRING DEFAULT '03-gold';

    -- Defining script variables
    DECLARE OR REPLACE create_statement STRING DEFAULT '';


    -- Create a new catalog for the project
    SET VAR create_statement = CONCAT(
            'CREATE CATALOG IF NOT EXISTS ',
            catalog_name
            );
            
    EXECUTE IMMEDIATE create_statement;

    -- Use the new catalog
    EXECUTE IMMEDIATE CONCAT('USE CATALOG ', catalog_name);

    -- Create initial schema to get raw data that will be tranfered from kennel 
    SET VAR create_statement = CONCAT(
            'CREATE SCHEMA IF NOT EXISTS ',
            schema_name_raw
            );

    EXECUTE IMMEDIATE create_statement;

        -- Create bronze schema to store 
    SET VAR create_statement = CONCAT(
            'CREATE SCHEMA IF NOT EXISTS ',
            schema_name_bronze
            );

    EXECUTE IMMEDIATE create_statement;

        -- Create silver schema to store  
    SET VAR create_statement = CONCAT(
            'CREATE SCHEMA IF NOT EXISTS ',
            schema_name_silver
            );

    EXECUTE IMMEDIATE create_statement;

        -- Create gold schema to store analytic data (data warehose) 
    SET VAR create_statement = CONCAT(
            'CREATE SCHEMA IF NOT EXISTS ',
            schema_name_gold
            );

    EXECUTE IMMEDIATE create_statement;



    
```
