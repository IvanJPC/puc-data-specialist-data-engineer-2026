# Catalog and Schemas creation
The Medallion Architecture is a data lakehouse design pattern that organizes data into layers based on quality and refinement level. The project will use this pattern

- Raw Layer (00-raw)

    Purpose: Initial landing zone for data exactly as ingested from source systems

    Characteristics:

        Unprocessed, unvalidated data
        Original structure and format preserved
        Immutable record of source data
        Includes metadata like extraction timestamp and dataset source
        A list of volumes will receive the source data transfered and preserved in order to be refined through each layer  
    Done

- Bronze Layer (01-bronze)

    Purpose: Raw data converted to Delta Lake format with basic structure

    Characteristics:

        Minimal transformations (e.g., type casting, schema enforcement, meaningful table names)
        Append-only or change data capture patterns
        Data quality issues may still exist
        Full history retained for auditability
        Includes metadata like ingestion timestamp and sources
    Done

- Silver Layer (02-silver)

    Purpose: Cleaned, validated, and enriched data

    Characteristics:

        Deduplication, data quality rules applied
        Business logic transformations
        Joins across sources
        "Single source of truth" for entities
        May use SCD (Slowly Changing Dimensions) patterns

- Gold Layer (03-gold)

    Purpose: Business-level aggregates and analytics-ready datasets

    Characteristics:

        Optimized for specific business use cases
        Pre-aggregated metrics, dashboards, reports
        Denormalized for query performance
        Domain-specific data models (data warehouse tables)
        

```python
    %sql
    -- Schema names using number prefix to mantain logical ordering by name
    
    -- Droping whole catalog for catalog refresh
    DROP CATALOG IF EXISTS puc_data_specialist_de_2026_09 CASCADE;

    -- Create a new catalog for the project
    CREATE CATALOG IF NOT EXISTS puc_data_specialist_de_2026_09;
    COMMENT ON CATALOG puc_data_specialist_de_2026_09 IS
      'PUC-RJ Data Specialist program - Data Egineering project catalog implementing Medallion Architecture (Raw, Bronze, Silver, Gold).';
            
    -- Use the new catalog
    USE CATALOG puc_data_specialist_de_2026_09;


    -- Create initial schema to get raw data that will be tranfered from kennel 
    CREATE SCHEMA IF NOT EXISTS `00-raw`;
    COMMENT ON SCHEMA `00-raw` IS
      'Initial landing zone for data exactly as ingested from source systems. Unprocessed, unvalidated data preserving original structure and format. Includes extraction metadata like timestamps and dataset source. This schema will only store raw dataset extracted from the source and stored into a Volume. 
      Dataset source: https://www.kaggle.com/datasets/olistbr/brazilian-ecommerce';

    -- Create bronze schema to store Raw data converted to Delta Lake format with basic structure
    CREATE SCHEMA IF NOT EXISTS `01-bronze`;
    COMMENT ON SCHEMA `01-bronze` IS
      'Raw data converted to Delta Lake format with basic structure. Minimal transformations (type casting, schema enforcement, meaningful table names). Full history retained for auditability. Includes ingestion timestamps and source metadata.';

    
    -- Create silver schema to store cleaned, validated, and enriched data 
    CREATE SCHEMA IF NOT EXISTS `02-silver`;
    COMMENT ON SCHEMA `02-silver` IS
      'Cleaned, validated, and enriched data. Deduplication and data quality rules applied. Business logic transformations and joins across sources. Single source of truth for entities. May use SCD (Slowly Changing Dimensions) patterns.';

    -- Create gold schema to store analytic-ready data (data warehose) 
    CREATE SCHEMA IF NOT EXISTS `03-gold`;
    COMMENT ON SCHEMA `03-gold` IS
      'Business-level aggregates and analytics-ready datasets. Optimized for specific business use cases. Pre-aggregated metrics, dashboards, and reports. Denormalized for query performance. Domain-specific data warehouse tables.';

    

    
```
